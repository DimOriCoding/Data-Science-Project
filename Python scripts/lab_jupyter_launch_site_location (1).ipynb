```python
# TASK 3: Calculate the distances between a launch site to its proximities
import numpy as np
def haversine(lat1, lon1, lat2, lon2):
    """
    Calculate the great-circle distance between two points on the Earth.
    Inputs in decimal degrees, output in kilometers.
    """
    R = 6371  # Earth radius in km
    lat1, lon1, lat2, lon2 = map(np.radians, [lat1, lon1, lat2, lon2])
    dlat = lat2 - lat1
    dlon = lon2 - lon1
    a = np.sin(dlat/2)**2 + np.cos(lat1)*np.cos(lat2)*np.sin(dlon/2)**2
    c = 2 * np.arcsin(np.sqrt(a))
    return R * c

spacex_df_unique = spacex_df[['Launch Site', 'Lat', 'Long']].drop_duplicates()
sites = spacex_df_unique
n = len(sites)
dist_matrix = np.zeros((n, n))

for i in range(n):
    for j in range(n):
        dist_matrix[i, j] = haversine(
            sites.iloc[i]['Lat'], sites.iloc[i]['Long'],
            sites.iloc[j]['Lat'], sites.iloc[j]['Long']
        )

# Convert to DataFrame for readability
distance_df = pd.DataFrame(dist_matrix, index=sites['Launch Site'], columns=sites['Launch Site'])
distance_df
```

Now zoom in to a launch site and explore its proximity to see if you can easily find any railway, highway, coastline, etc. Move your mouse to these points and mark down their coordinates (shown on the top-left) in order to the distance to the launch site.


```python
from math import sin, cos, sqrt, atan2, radians

def calculate_distance(lat1, lon1, lat2, lon2):
    # approximate radius of earth in km
    R = 6373.0

    lat1 = radians(lat1)
    lon1 = radians(lon1)
    lat2 = radians(lat2)
    lon2 = radians(lon2)

    dlon = lon2 - lon1
    dlat = lat2 - lat1

    a = sin(dlat / 2)**2 + cos(lat1) * cos(lat2) * sin(dlon / 2)**2
    c = 2 * atan2(sqrt(a), sqrt(1 - a))

    distance = R * c
    return distance
```

```python
# Create a marker with distance to a closest city, railway, highway, etc.
# Draw a line between the marker to the launch site
def add_distance_marker(site_map, launch_lat, launch_lon, target_lat, target_lon, label, line_color='blue'):
    distance = calculate_distance(launch_lat, launch_lon, target_lat, target_lon)
    
    target_coordinate = [target_lat, target_lon]
    launch_coordinate = [launch_lat, launch_lon]
    
    # marker on the selected target point
    folium.Marker(
        location=target_coordinate,
        popup=f'{label} ({distance:.2f} km)'
    ).add_to(site_map)
    
    # text label showing distance
    folium.Marker(
        location=target_coordinate,
        icon=DivIcon(
            icon_size=(20, 20),
            icon_anchor=(0, 0),
            html=f'<div style="font-size: 12px; color:#d35400;"><b>{label}: {distance:.2f} KM</b></div>'
        )
    ).add_to(site_map)
    
    # line from launch site to target point
    folium.PolyLine(
        locations=[launch_coordinate, target_coordinate],
        weight=2,
        color=line_color
    ).add_to(site_map)
    
    return distance
```

```python
# launch site coordinates (example: CCAFS LC-40)
launch_lat = 28.562302
launch_lon = -80.577356

# replace these with the points you selected on the map
closest_city_lat = 28.0836
closest_city_lon = -80.6081

closest_railway_lat = 28.5721
closest_railway_lon = -80.5853

closest_highway_lat = 28.5632
closest_highway_lon = -80.5708

# add markers + lines
distance_city = add_distance_marker(
    site_map, launch_lat, launch_lon,
    closest_city_lat, closest_city_lon,
    'Closest City', 'green'
)

distance_railway = add_distance_marker(
    site_map, launch_lat, launch_lon,
    closest_railway_lat, closest_railway_lon,
    'Closest Railway', 'red'
)

distance_highway = add_distance_marker(
    site_map, launch_lat, launch_lon,
    closest_highway_lat, closest_highway_lon,
    'Closest Highway', 'purple'
)

site_map
```
